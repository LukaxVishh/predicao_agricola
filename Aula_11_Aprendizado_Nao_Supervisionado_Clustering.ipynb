# 📘 MÓDULO 1: Do Aprendizado Supervisionado ao Não Supervisionado

## 1.1. A Fronteira Epistemológica: O Que Acontece Quando Não Temos o "Gabarito" ($y$)?
Na **Aula 10**, treinamos algoritmos de **Aprendizado Supervisionado Preditivo (Regressão Linear)**. Naquele cenário, fornecíamos ao computador uma matriz de variáveis explicativas ($X$) acompanhada de uma variável alvo histórica conhecida ($y$, o "supervisor" ou gabarito, como o `Faturamento_Mensal` da loja).

Mas imagine que a diretoria comercial entregue a você uma base com milhares de clientes contendo `Renda Mensal`, `Frequência de Visitas` e `Tempo de Cadastro`, **sem nenhuma coluna dizendo a qual categoria cada cliente pertence**, e faça a seguinte pergunta:  
> *"Quais são os grupos ou tribos naturais de clientes que existem dentro da nossa operação?"*

Entramos no território do **Aprendizado Não Supervisionado (*Unsupervised Learning*)**:
* **Ausência de Variável Alvo ($y$):** Trabalhamos unicamente com a matriz de atributos observados $X$.
* **Objetivo do *Clustering* (Agrupamento):** Descobrir padrões e estruturas latentes nos dados, particionando as observações em grupos (*clusters*) que respeitem dois princípios geométricos simultâneos (Jain, Murty & Flynn, *ACM Computing Surveys*, 1999):
  1. **Alta Coesão Interna (*Intra-cluster similarity*):** Indivíduos dentro de um mesmo grupo devem ser o mais semelhantes possível entre si.
  2. **Alto Acoplamento/Separação Externa (*Inter-cluster dissimilarity*):** Indivíduos de grupos distintos devem ser o mais diferentes possível entre si.

| Dimensão | 🎯 Aprendizado Supervisionado (Aula 10 - Regressão) | 🧭 Aprendizado Não Supervisionado (Aula 11 - Clustering) |
| :--- | :--- | :--- |
| **Estrutura dos Dados** | Dados rotulados $(X, y)$ — exige variável resposta $y$ (gabarito). | Dados não rotulados $(X)$ — **não existe variável resposta $y$**. |
| **Pergunta de Negócio** | *"Quanto este cliente específico vai gastar amanhã?"* ($\hat{y}$) | *"Quais são os perfis comportamentais que existem na minha carteira?"* |
| **Métrica de Sucesso** | Desvio entre previsão e realidade ($MAE$, $RMSE$, $R^2$). | Compacidade e separação geométrica (*Inércia / WCSS*, *Silhueta*). |

> 🏛️ **Contexto Histórico & Etimológico:**  
> * **Etimologia:** A palavra *Cluster* vem do inglês antigo *clyster* ("feixe", "cacho de uvas", "aglomerado"). O termo *Cluster Analysis* foi introduzido na literatura científica pelos psicometristas **Joseph Zubin (1938)** e **Robert Tryon (1939)** para agrupar perfis psicológicos.  
> * **A Paternidade Tripla do K-Means:** O algoritmo K-Means foi descoberto independentemente três vezes: a formulação geométrica nasceu com o matemático polonês **Hugo Steinhaus (1956)**; o algoritmo iterativo que usamos hoje foi criado em **1957 por Stuart Lloyd** nos laboratórios da *Bell Labs* para compressão de sinais de telecomunicações (*Pulse-Code Modulation*), mas ficou retido por questões jurídicas de patentes até 1982; e o batismo do nome **"K-Means"** foi feito por **James MacQueen (1967)** na UCLA.

## 1.2. A Geometria da Proximidade: Distância Euclidiana

Na Regressão Linear, medíamos o erro apenas na vertical ($e_i = y_i - \hat{y}_i$). No *Clustering*, como todas as colunas pertencem à matriz $X$, o algoritmo mede a "semelhança" entre dois indivíduos $p$ e $q$ calculando a distância em linha reta entre eles no espaço multidimensional — a **Distância Euclidiana** (Teorema de Pitágoras generalizado para $m$ variáveis):

$$d(p, q) = \sqrt{(x_{p1} - x_{q1})^2 + (x_{p2} - x_{q2})^2 + \dots + (x_{pm} - x_{qm})^2} = \sqrt{\sum_{j=1}^{m} (x_{pj} - x_{qj})^2}$$

<small>
📖 <b>Guia de Leitura Matemática e Notação (População vs. Amostra):</b><br>
• <b>$X$ vs. $x_{ij}$ (Lê-se <i>"xis"</i> — letra latina, não confundir com a letra grega <i>Qui</i> $\chi$ usada no teste $\chi^2$):</b> Na estatística, letras maiúsculas ($X$) representam a matriz completa ou a variável aleatória na <b>População</b>, enquanto letras minúsculas com subscritos ($x_{ij}$, lê-se <i>"xis índice i, jota"</i>) representam o valor específico observado na <b>Amostra</b> para o indivíduo (linha) $i$ na variável (coluna) $j$.<br>
• <b>$d(p, q)$ (Lê-se <i>"dê de pê, quê"</i>):</b> Distância euclidiana entre o indivíduo amostral $p$ (linha $p$) e o indivíduo amostral $q$ (linha $q$).<br>
• <b>$\sum_{j=1}^{m}$ (Letra grega <i>Sigma</i> maiúscula — lê-se <i>"somatório de jota igual a 1 até éme"</i>):</b> Indica que devemos somar as diferenças ao quadrado coluna por coluna, desde a primeira variável ($j=1$) até a última variável ($m$).
</small>

## 1.3. Como o Algoritmo K-Means Funciona? (Os 4 Passos de Lloyd e a Inércia)

O **K-Means** agrupa os dados executando um ciclo iterativo de 4 etapas:
1. **Passo 1 (Definição de $K$ e Sorteio Inicial):** Você define o número $K$ de grupos desejados (ex: $K = 3$). O algoritmo posiciona $K$ pontos iniciais no espaço chamados de **Centroides ($\mu_1, \mu_2, \dots, \mu_K$)**.
2. **Passo 2 (Atribuição / *Assignment*):** Para cada linha $x_i$ da tabela, o algoritmo calcula a Distância Euclidiana até cada um dos $K$ centroides e atribui o cliente ao **centroide mais próximo**.
3. **Passo 3 (Atualização do Centro de Gravidade / *Update*):** Com os grupos formados, cada centroide $\mu_k$ se desloca para a **média aritmética exata** de todos os pontos que foram atribuídos ao seu grupo (daí o nome *K-Means* — $K$-Médias).
4. **Passo 4 (Convergência):** Os Passos 2 e 3 se repetem até que nenhum indivíduo mude de grupo e os centroides parem de se mover.

### A Função Objetivo do K-Means: Inércia (*WCSS*)
Matematicamente, o método `.fit()` do K-Means busca minimizar a **Inércia**, conhecida na literatura acadêmica como **WCSS** (*Within-Cluster Sum of Squares* — Soma dos Quadrados Intra-Cluster):

$$\text{Inércia (WCSS)} = \sum_{k=1}^{K} \sum_{x_i \in C_k} \|x_i - \mu_k\|^2$$

<small>
📖 <b>Guia de Leitura Matemática e Notação:</b><br>
• <b>$K$ vs. $k$ (Lê-se <i>"ká maiúsculo"</i> e <i>"ká minúsculo"</i>):</b> $K$ é o número total de clusters escolhidos; $k$ é o contador do cluster atual ($k = 1, 2, \dots, K$).<br>
• <b>$x_i \in C_k$ (Lê-se <i>"xis índice i pertencente a Cê índice ká"</i>):</b> Representa cada vetor de observação amostral $x_i$ que faz parte do conjunto de membros do cluster $C_k$.<br>
• <b>$\mu_k$ (Letra grega <i>Mi</i> minúscula — lê-se <i>"mi índice ká"</i>):</b> É o vetor de coordenadas do <b>Centroide</b> do cluster $k$. <i>(Nota de rigor estatístico: na inferência clássica, $\mu$ designa a média da <b>População</b> e $\bar{x}$ — lê-se "xis barra" — designa a média da <b>Amostra</b>. Em Machine Learning, usa-se $\mu_k$ por convenção histórica de Lloyd/MacQueen para representar o baricentro amostral do cluster $k$).</i><br>
• <b>$\|x_i - \mu_k\|^2$ (Lê-se <i>"norma euclidiana ao quadrado de xis i menos mi ká"</i>):</b> É o quadrado da distância euclidiana entre o ponto amostral $x_i$ e o centroide $\mu_k$ do seu grupo.
</small>

# 🧹 MÓDULO 2: Caso Prático Visível (20 Clientes) e a "Armadilha da Escala" (`StandardScaler`)

## 2.1. Nosso Cenário Didático: Programa de Fidelidade "Café com Dados" (20 Clientes)
Assim como fizemos nas Aulas 09 e 10, vamos começar com uma tabela enxuta de **20 clientes (`Cli_01` a `Cli_20`)** para podermos auditar visualmente cada linha na tela do projetor.

Temos duas variáveis numéricas descrevendo o comportamento de cada cliente ($X$):
* `Renda_Mensal_BRL`: Renda mensal declarada pelo cliente (varia de **R$ 2.200 a R$ 18.500**).
* `Visitas_Mensais`: Quantidade de vezes que o cliente visitou a cafeteria no último mês (varia de **2 a 28 visitas**).

Inspecione a tabela gerada na célula abaixo. Veja se você consegue enxergar as **3 tribos comportamentais** que plantamos nos dados:
1. **`Cli_01` a `Cli_07`:** Renda mais baixa (~R$ 2.600), mas visitam a cafeteria quase todo dia (**23 a 28 visitas/mês**).
2. **`Cli_08` a `Cli_14`:** Renda intermediária (~R$ 8.100), mas raramente aparecem (**2 a 5 visitas/mês**).
3. **`Cli_15` a `Cli_20`:** Renda alta (~R$ 16.700) e frequência elevada (**16 a 21 visitas/mês**).

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score

pd.set_option('display.max_rows', 25)
pd.set_option('display.width', 1000)
plt.rcParams['figure.figsize'] = (14, 5.5)
sns.set_theme(style='whitegrid')

# Base didática de 20 clientes (Sem variável resposta y -> Aprendizado Não Supervisionado!)
dados_clientes = {
    'ID_Cliente': [f'Cli_{i:02d}' for i in range(1, 21)],
    'Renda_Mensal_BRL': [
        # Tribo A (Cli_01 a Cli_07): Renda baixa, altíssima frequência
        2200, 2400, 2500, 2700, 2800, 3000, 3100,
        # Tribo B (Cli_08 a Cli_14): Renda média, baixíssima frequência
        7200, 7500, 7800, 8100, 8400, 8800, 9100,
        # Tribo C (Cli_15 a Cli_20): Renda alta, alta frequência
        15200, 15800, 16400, 17000, 17800, 18500
    ],
    'Visitas_Mensais': [
        24, 26, 25, 28, 23, 27, 25,
        2,  4,  3,  2,  5,  3,  4,
        16, 18, 17, 20, 19, 21
    ]
}

df_clientes = pd.DataFrame(dados_clientes)
print(f"Dimensões da matriz X: {df_clientes.shape[0]} clientes x 2 variáveis numéricas")
display(df_clientes)
```

---
## 2.2. O "Sequestro da Distância Euclidiana" pela Escala e o `StandardScaler`

O que acontece se jogarmos `Renda_Mensal_BRL` e `Visitas_Mensais` diretamente no K-Means sem tratar a escala das variáveis?  
Olhe para a conta que a Distância Euclidiana fará entre o **`Cli_07`** ($\text{Renda} = 3.100, \text{Visitas} = 25$) e o **`Cli_08`** ($\text{Renda} = 7.200, \text{Visitas} = 2$):

$$d(\text{Cli}_{07}, \text{Cli}_{08}) = \sqrt{\underbrace{(3.100 - 7.200)^2}_{16.810.000} + \underbrace{(25 - 2)^2}_{529}} = \sqrt{16.810.529}$$

Repare no absurdo numérico: uma diferença de R$ 4.100 na renda gera **16,8 milhões** dentro da raiz quadrada, enquanto uma diferença abissal de **23 visitas mensais** gera míseros **529**!  
Ou seja: sem padronização, **a variável de maior grandeza numérica (R$) sequestra 99,997% da Distância Euclidiana**, tornando o K-Means completamente cego para a coluna `Visitas_Mensais`!

### A Solução Obrigatória: Padronização Z-Score (`StandardScaler`)
Para que todas as variáveis tenham exatamente o mesmo peso geométrico, aplicamos a transformação **Z-Score**, fazendo com que cada coluna passe a ter **Média $= 0$** e **Desvio-Padrão $= 1$**:

$$z_{ij} = \frac{x_{ij} - \bar{x}_j}{s_j} \quad \left(\text{ou na formulação populacional do Scikit-Learn: } z_{ij} = \frac{x_{ij} - \mu_j}{\sigma_j}\right)$$

<small>
📖 <b>Guia de Leitura Matemática e Diferença entre População e Amostra:</b><br>
• <b>$z_{ij}$ (Lê-se <i>"zê índice i, jota"</i> ou <i>Escore-Z</i>):</b> Valor padronizado (adimensional, sem unidade em R$ ou visitas) da amostra $i$ na coluna $j$. Indica a quantos desvios-padrão aquele indivíduo está acima ($z > 0$) ou abaixo ($z < 0$) da média da coluna.<br>
• <b>$\bar{x}_j$ vs. $\mu_j$ (Média Amostral vs. Média Populacional):</b> $\bar{x}_j = \frac{1}{n}\sum_{i=1}^{n} x_{ij}$ (lê-se <i>"xis barra índice jota"</i>) é a média calculada na <b>Amostra</b> de tamanho $n$; já $\mu_j$ (letra grega <i>Mi</i>, lê-se <i>"mi índice jota"</i>) representa a média verdadeira da <b>População</b>.<br>
• <b>$s_j$ vs. $\sigma_j$ (Desvio-Padrão Amostral vs. Populacional):</b> $s_j = \sqrt{\frac{1}{n-1}\sum (x_{ij} - \bar{x}_j)^2}$ (lê-se <i>"ésse índice jota"</i>) é o desvio-padrão da <b>Amostra</b> (dividido por $n-1$, Correção de Bessel); já $\sigma_j$ (letra grega <i>Sigma</i> minúscula, lê-se <i>"sigma índice jota"</i>) usa divisor $n$ (<b>População</b>). <i>Nota técnica: O <code>StandardScaler</code> do Scikit-Learn divide por $\sigma_j$ (divisor $n$), garantindo variância unitária exata em todas as colunas!</i>
</small>

Vamos provar visualmente na célula abaixo a diferença entre rodar o K-Means **sem** e **com** `StandardScaler`!

```python
X_clientes = df_clientes[['Renda_Mensal_BRL', 'Visitas_Mensais']].copy()

# 1. K-Means BRUTO (Sem StandardScaler) -> Olha quase exclusivamente para a Renda!
km_sem_escala = KMeans(n_clusters=3, random_state=42, n_init=10)
df_clientes['Cluster_Sem_Escala'] = km_sem_escala.fit_predict(X_clientes)

# 2. Aplicando o StandardScaler (Z-Score)
scaler = StandardScaler()
X_clientes_scaled = scaler.fit_transform(X_clientes)

df_zscore = pd.DataFrame(X_clientes_scaled, columns=['Renda_ZScore (z1)', 'Visitas_ZScore (z2)'], index=df_clientes['ID_Cliente'])
print("🔍 Veja como ficam as primeiras 5 linhas após o StandardScaler (Média = 0, Desvio Padrão = 1):")
display(df_zscore.head(5).round(3))

# 3. K-Means PADRONIZADO (Com StandardScaler)
km_padronizado = KMeans(n_clusters=3, random_state=42, n_init=10)
df_clientes['Cluster_Padronizado'] = km_padronizado.fit_predict(X_clientes_scaled)

# Devolvendo os centroides padronizados para a escala original (R$ e Visitas) apenas para plotar no gráfico
centroides_reais = scaler.inverse_transform(km_padronizado.cluster_centers_)

fig, axes = plt.subplots(1, 2, figsize=(15, 5.5))

# Painel 1: Sem Escala (Fronteiras verticais cegas para o eixo Y)
sns.scatterplot(
    data=df_clientes, x='Renda_Mensal_BRL', y='Visitas_Mensais',
    hue='Cluster_Sem_Escala', palette='Set1', s=110, ax=axes[0]
)
axes[0].scatter(
    km_sem_escala.cluster_centers_[:, 0], km_sem_escala.cluster_centers_[:, 1],
    color='black', marker='X', s=220, label='Centroides Brutos'
)
axes[0].set_title('1. K-Means SEM Padronização (Dominado pela grandeza do R$)')
axes[0].set_xlabel('Renda Mensal (R$)')
axes[0].set_ylabel('Visitas Mensais')
axes[0].legend()

# Painel 2: Com StandardScaler (Geometria equilibrada entre X e Y)
sns.scatterplot(
    data=df_clientes, x='Renda_Mensal_BRL', y='Visitas_Mensais',
    hue='Cluster_Padronizado', palette='viridis', s=110, ax=axes[1]
)
axes[1].scatter(
    centroides_reais[:, 0], centroides_reais[:, 1],
    color='red', marker='X', s=220, zorder=5, label='Centroides Padronizados (μ_k)'
)
for _, r in df_clientes.iterrows():
    axes[1].annotate(r['ID_Cliente'].replace('Cli_', 'C'), (r['Renda_Mensal_BRL']+180, r['Visitas_Mensais']-0.6), fontsize=8.5)
axes[1].set_title('2. K-Means COM StandardScaler (Pesos Equalizados via Z-Score)')
axes[1].set_xlabel('Renda Mensal (R$)')
axes[1].set_ylabel('Visitas Mensais')
axes[1].legend()

plt.tight_layout()
plt.show()
```

### 🔥 Teste de Fogo: E se dois clientes tiverem a MESMA renda, mas comportamentos opostos de visita?
Para ver o erro do K-Means sem escala de forma incontestável, imagine que entra na base um **`Cli_21_Teste`** com **Renda de R$ 8.000** (idêntica à da Tribo B), mas que visita a loja **27 vezes por mês** (comportamento típico dos Fãs Diários da Tribo A)!  
Execute a célula abaixo e veja onde cada modelo aloca o `Cli_21_Teste`:

```python
# Adicionando o Cli_21_Teste: Renda Média (R$ 8.000), mas Altíssima Frequência (27 visitas/mês!)
df_teste_escala = pd.concat([
    df_clientes[['ID_Cliente', 'Renda_Mensal_BRL', 'Visitas_Mensais']],
    pd.DataFrame({'ID_Cliente': ['Cli_21_Teste'], 'Renda_Mensal_BRL': [8000], 'Visitas_Mensais': [27]})
], ignore_index=True)

# 1. Sem StandardScaler:
km_errado = KMeans(n_clusters=3, random_state=42, n_init=10).fit(df_teste_escala[['Renda_Mensal_BRL', 'Visitas_Mensais']])
df_teste_escala['Grupo_SEM_Scaler'] = km_errado.labels_

# 2. Com StandardScaler:
X_teste_sc = StandardScaler().fit_transform(df_teste_escala[['Renda_Mensal_BRL', 'Visitas_Mensais']])
km_certo = KMeans(n_clusters=3, random_state=42, n_init=10).fit(X_teste_sc)
df_teste_escala['Grupo_COM_Scaler'] = km_certo.labels_

print("Compare onde o Cli_21_Teste (R$ 8.000 e 27 visitas) foi alocado em relação ao Cli_06 (Fã Diário: 27 visitas) e ao Cli_08 (Raro: 2 visitas):")
display(df_teste_escala.iloc[[5, 7, 20]])
print("💡 Diagnóstico:")
print("   -> SEM Scaler: O K-Means ignora as 27 visitas e joga o Cli_21 no mesmo grupo do Cli_08 (2 visitas) só por causa dos R$ 8.000!")
print("   -> COM StandardScaler: O K-Means pondera Renda e Visitas igualmente e agrupa o Cli_21 junto com os Fãs Diários (Cli_06)!")
```

# 📈 MÓDULO 3: Descobrindo o Número Ideal de Grupos ($K$) e Criando *Personas*

Nos exemplos anteriores, nós já sabíamos de antemão que existiam $K = 3$ grupos. Mas em uma base de dados real com 6 ou 10 variáveis simultâneas, **como descobrimos qual é o número $K$ ideal de clusters?**

Utilizamos duas métricas de diagnóstico:

### 3.1. O Método do Cotovelo (*Elbow Method* — Inércia / WCSS)
Testamos vários valores de $K$ (ex: de $K=2$ até $K=7$) e plotamos a **Inércia ($WCSS$)** no eixo vertical.  
Conforme $K$ cresce, a Inércia sempre diminui. Procuramos no gráfico o ponto onde a curva forma um **"cotovelo"** (o ponto de inflexão onde adicionar mais um grupo traz uma redução marginal pequena na inércia).

### 3.2. O Coeficiente de Silhueta (*Silhouette Score* — Rousseeuw, 1987)
Para não depender apenas de interpretação visual subjetiva, calculamos o **Coeficiente de Silhueta**, que mede a qualidade geométrica da separação para cada ponto amostral $i$:

$$S(i) = \frac{b(i) - a(i)}{\max\{a(i), b(i)\}}$$

<small>
📖 <b>Guia de Leitura Matemática e Notação:</b><br>
• <b>$S(i)$ (Lê-se <i>"éssi de i"</i>):</b> Coeficiente de Silhueta da observação amostral $i$. Varia no intervalo $[-1, +1]$. Na avaliação global do modelo, calculamos a média amostral $\bar{S}$ de todos os indivíduos.<br>
• <b>$a(i)$ (Lê-se <i>"á de i"</i> — Coesão Intra-cluster):</b> Distância média entre o indivíduo amostral $i$ e todos os demais indivíduos que pertencem ao <b>mesmo cluster</b> que ele (quanto menor $a(i)$, mais compacto é o grupo).<br>
• <b>$b(i)$ (Lê-se <i>"bê de i"</i> — Separação Inter-cluster):</b> Menor distância média entre o indivíduo amostral $i$ e todos os indivíduos do <b>cluster vizinho mais próximo</b> ao qual ele não pertence (quanto maior $b(i)$, mais bem separado está o grupo dos demais).<br>
• <b>$\max\{a(i), b(i)\}$ (Lê-se <i>"máximo entre á de i e bê de i"</i>):</b> Fator de normalização no denominador que garante que $S(i)$ fique sempre entre $-1$ (ponto no cluster errado) e $+1$ (clusterização perfeita).
</small>

```python
# Testando K de 2 até 7 nos nossos 20 clientes padronizados para encontrar o K ótimo!
lista_inercia = []
lista_silhueta = []
faixa_k = range(2, 8)

for k in faixa_k:
    modelo_k = KMeans(n_clusters=k, random_state=42, n_init=10)
    rotulos_k = modelo_k.fit_predict(X_clientes_scaled)
    lista_inercia.append(modelo_k.inertia_)
    lista_silhueta.append(silhouette_score(X_clientes_scaled, rotulos_k))

# Exibindo a tabela numérica de diagnóstico
df_diagnostico_k = pd.DataFrame({
    'K (Nº de Clusters)': list(faixa_k),
    'Inércia (WCSS - Menor é mais compacto)': np.round(lista_inercia, 3),
    'Silhueta Média S (Maior é mais bem separado)': np.round(lista_silhueta, 4)
})
display(df_diagnostico_k)

fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))

# Gráfico 1: Método do Cotovelo
axes[0].plot(faixa_k, lista_inercia, 'bo-', lw=2.2, markersize=8)
axes[0].axvline(3, color='crimson', linestyle='--', lw=2, label='Cotovelo Nítido em K = 3')
axes[0].set_title('1. Método do Cotovelo (Queda da Inércia / WCSS)')
axes[0].set_xlabel('Número de Clusters (K)')
axes[0].set_ylabel('Inércia (WCSS)')
axes[0].legend()

# Gráfico 2: Coeficiente de Silhueta
axes[1].plot(faixa_k, lista_silhueta, 'go-', lw=2.2, markersize=8)
axes[1].axvline(3, color='crimson', linestyle='--', lw=2, label='Pico Máximo em K = 3 (S = 0.88)')
axes[1].set_title('2. Coeficiente de Silhueta Médio (Pico = Melhor K)')
axes[1].set_xlabel('Número de Clusters (K)')
axes[1].set_ylabel('Silhouette Score Médio')
axes[1].legend()

plt.tight_layout()
plt.show()
```

## 3.3. Do Número Frio à Estratégia Executiva: Batizando as *Personas* dos Clusters
O algoritmo `KMeans` devolve apenas códigos numéricos (`Cluster 0`, `Cluster 1`, `Cluster 2`). Apresentar apenas números para um diretor de negócios não gera ação.  
O papel analítico do Cientista de Dados no Aprendizado Não Supervisionado é agregar as médias amostrais reais ($\bar{x}$) de cada grupo com `.groupby()` e traduzir os números em **Personas de Negócio**:

```python
# Calculando as estatísticas descritivas amostrais (na escala original em R$ e Visitas) para cada Cluster
perfil_executivos = df_clientes.groupby('Cluster_Padronizado').agg(
    Quantidade_Clientes=('ID_Cliente', 'count'),
    Renda_Media_Amostral_BRL=('Renda_Mensal_BRL', 'mean'),
    Desvio_Padrao_Renda_BRL=('Renda_Mensal_BRL', 'std'),
    Visitas_Medias_Mes=('Visitas_Mensais', 'mean')
).round(1)

# Traduzindo as médias de cada cluster em Personas de Negócio e Planos de Ação
nomes_personas = []
acoes_recomendadas = []
for idx, row in perfil_executivos.iterrows():
    if row['Renda_Media_Amostral_BRL'] > 12000:
        nomes_personas.append('🌟 VIPs / Executivos (Alta Renda & Alta Frequência)')
        acoes_recomendadas.append('Oferecer reserva de sala de reunião e cafés especiais em grãos premium.')
    elif row['Visitas_Medias_Mes'] > 15:
        nomes_personas.append('☕ Fãs Diários / Recorrentes (Ticket Moderado & Altíssima Visita)')
        acoes_recomendadas.append('Criar assinatura mensal de café expresso e combo matinal de fidelidade.')
    else:
        nomes_personas.append('💤 Visitantes Ocasionalmente Raros (Renda Média & Baixo Engajamento)')
        acoes_recomendadas.append('Disparar cupom de cashback no WhatsApp para visitas em dias de semana.')

perfil_executivos['Persona_de_Negocio'] = nomes_personas
perfil_executivos['Acao_Estrategica_Sugerida'] = acoes_recomendadas

print("📋 TABELA EXECUTIVA DE SEGMENTAÇÃO DE CLIENTES (ENTREGÁVEL DE NEGÓCIO):")
display(perfil_executivos)
```

## 4.2. Recapitulação da Matemática da Nota e Avaliação por Pares (Soma Zero)

A nota base do Projeto no 2º Bimestre é de **$5,0$ pontos** ($2,5$ pontos divididos nos 2 Checkpoints de desenvolvimento + $2,5$ pontos na Entrega Final em Vídeo Pitch e Notebook no Blackboard até 6 dias corridos após a última aula).

$$\mathbf{Nota\_Final_{2B}} = \text{Nota\_Base} \times M_{real} \times F_{ind}$$

Onde o **Fator de Mérito Individual ($F_{ind}$)** de cada aluno $i$ é calculado a partir das notas atribuídas pelos seus $(N - 1)$ colegas de equipe em um jogo de **Soma Zero** (onde cada avaliador distribui exatos $5 \times (N - 1)$ pontos entre seus pares):

$$F_{ind} = 0,5 + 0,1 \times \bar{m}_i$$

<small>
📖 <b>Guia de Leitura Matemática dos Termos da Avaliação:</b><br>
• <b>$M_{real}$ (Lê-se <i>"éme índice real"</i> — Multiplicador de Mundo Real):</b> Vale $\mathbf{1,2}$ se o grupo trouxer dados operacionais reais de uma empresa/negócio (permitindo atingir até $6,0$ pontos no projeto!) ou $\mathbf{1,0}$ se utilizar bases públicas (Kaggle/UCI/Governo) ou sintéticas.<br>
• <b>$N$ (Lê-se <i>"êne maiúsculo"</i>):</b> Tamanho total da equipe ($3$ ou $4$ integrantes). Cada aluno avalia apenas seus $(N - 1)$ colegas.<br>
• <b>$\bar{m}_i$ (Lê-se <i>"éme barra índice i"</i>):</b> É a <b>média amostral</b> dos pontos que o aluno $i$ recebeu dos seus $(N - 1)$ colegas de grupo ($\bar{m}_i = \frac{1}{N-1}\sum m_{ji}$). Se todos trabalharam igual, $\bar{m}_i = 5 \implies F_{ind} = 1,0\times$. Se um integrante não fez nada ($\bar{m}_i = 0$), cai para $F_{ind} = 0,5\times$ e os pontos que ele perdeu elevam a média $\bar{m}$ dos colegas produtivos (podendo chegar a $F_{ind} = 1,25\times$).
</small>
